```python
raw_data_folder = Path.cwd().parent/"data"/"raw"/"ligue1"
```

```python
seasons = {}
cols_set = []

for file in raw_data_folder.glob("*.csv"):
    df = pd.read_csv(file)
    seasons[file.stem] = df
    cols_set.append(set(df.columns))
    print(f"Found {file.stem} with {len(df.columns)} and {len(df)} rows")
```

```python
common_cols = set.intersection(*cols_set)
```
